```python
n = int(input("Digite um número inteiro"))
if n % 3 == 0 and n % 5 == 0:
    print ("O número" ,n, "é divisível por 3 e por 5")
else :
    print ("O número" ,n, "não é divisível por 3 por e 5")
```

```python
valor_compra = float(input("Digite o valor da compra para verificar qual é o desconto"))
if valor_compra <=  1000:
    percentual=10
    desconto = valor_compra * 0.10
elif valor_compra<= 5000:
    percentual = 20
    desconto = valor_compra * 0.20
else:
    percentual = 30
    desconto = valor_compra * 0.30

valor_final = valor_compra - desconto

print(f"\nValor da compra:  R$ {valor_compra:.2f}")
print(f"Desconto ({percentual}%):   R$ {desconto:.2f}")
print(f"Valor final:      R$ {valor_final:.2f}") 
    
    
     
```

```python
import math


largura = float(input("Informe a largura do aposento em metros: "))
comprimento = float(input("Comprimento do aposento em metros: "))
litros_lata = float(input("Quantidade de litros por lata: "))


PE_DIREITO = 2.80
AREA_PORTA = 0.80 * 2.10   
RENDIMENTO = 3             


area_paredes = (largura * PE_DIREITO * 2) + (comprimento * PE_DIREITO * 2)
area_total = area_paredes - AREA_PORTA


litros_necessarios = area_total / RENDIMENTO
latas = math.ceil(litros_necessarios / litros_lata)


print(f"\nÁrea total das paredes: {area_total:.2f} m²")
print(f"Litros de tinta necessários: {litros_necessarios:.2f} L")
print(f"Quantidade de latas: {latas}")
```
