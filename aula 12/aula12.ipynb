```python
cidades = []
while True:
    cidade = input ("Digite uma cidade <finalizar digite sair>:")
    if cidade.lower() == "sair":
        break
    cidades.append(cidade)

cidades.sort()
for cidade in cidades:
    print(cidade)
```

```python
from random import randint 
numeros = [0,0,0,0,0,0,0]
for i in range (6000):
    n = randint(1,6)
    numeros[n] = numeros[n] + 1
for i in range (1,7):
    p = (numeros [i]/6000)*100
    print(f"{i} - {numeros[i]} {p:6.2f}%")
    
```
