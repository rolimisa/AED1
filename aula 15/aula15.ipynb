1) Em um cercado existem vários patos e coelhos. 
Escreva uma função em linguagem Python que recebe como parâmetro o total de cabeças e o total de pés,
e retorne dois valores: o total de patos e o total de coelhos que se encontram nesse cercado. (4,0 pontos) 
Teste essa função elaborando um script que vai receber do usuário dois valores. Para tanto, leve em consideração que os dois últimos dígitos 
de seu RA sejam iguais a Y.  Entrada de Dados:    Total de Cabeças:  Y = _________                                      
Total de Pés: Y*3 = _________  Levando em consideração o resultado de seu script... Responda:  Total de Patos:___________________       
Total de Coelhos: ___________________ 

```python
def animais(cabecas,pes):
    
    coelhos = (pes - (2 * cabecas)) // 2
    patos = cabecas - coelhos
    return patos, coelhos

cabecas = int(input("Digite o total de cabeças"))
pes = int (input("Digite o total de pés"))

patos, coelhos = animais(cabecas,pes)

print ("O total de patos é: ",patos)
print ("O total de coelhos é: ",coelhos)
```

2) Faça um script em Python que recebe como entrada um número positivo no formato de String. 
Esse script deve devolver dois valores: o total da soma de todos os dígitos desse número e o valor da multiplicação de todos esses 
mesmos dígitos. Por exemplo: Se a entrada for 1234, o script deve devolver como resultado: Soma = 10 e Multiplicação = 24.      
(3,0 pontos) Passe o número completo de seu RA para o script, e responda:

```python
numero = input("Digite o número do seu RA")
soma = 0
multiplicacao = 1

for i in numero:
    soma = soma + int (i)
    multiplicacao = multiplicacao * int (i)
    numero +=i

print ("A soma é" ,soma)
print ("A multiplicação é", multiplicacao)

```

3) Reverso do número. 
Faça uma função que retorne o reverso de um número inteiro informado. Por exemplo: 127 -> 721.   ((use os números de seu RA para testar))

```python
def reverso(numero):
    numero_str = str(numero)
    numero_invertido = numero_str[::-1]
    return int(numero_invertido)

ra = int(input("Digite o número do seu RA: "))
print("Reverso:", reverso(ra))
```
