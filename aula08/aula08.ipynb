```python
import random

numero_sorteado= random.randint (1,50)
tentativas = 0 
limite = 5

while tentativas < limite:
        numero_escolhido = int (input("Digite um número"))
        tentativas += 1
        if numero_sorteado == numero_escolhido:
            print ("Parabéns")
            break
        else:
            if numero_escolhido > numero_sorteado:
                print ("O número escolhido",numero_escolhido, "é maior que o número sorteado")
            else:
                  print ("O número escolhido",numero_escolhido, "é menor que o número sorteado")
print ("Tentativas restantes",limite - tentativas)
            
if tentativas == limite and numero_escolhido != numero_sorteado:
            print ("Você perdeu! O número era", numero_sorteado)
```
